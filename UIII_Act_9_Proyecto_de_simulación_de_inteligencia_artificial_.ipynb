```python
# ==============================================
# IA DE YOUTUBER
# Nivel: Principiante
# Uso de:
# - listas
# - if
# - elif
# - else
# - while
# ==============================================

print("======================================")
print("     IA DE YOUTUBER VIRTUAL")
print(" CBTIS 128 - Programacion e IA")
print("======================================")

# Lista vacia al inicio
videos = []

opcion = 0

while opcion != 5:

    print("\n=========== MENU ===========")
    print("1. Agregar video")
    print("2. Ver videos")
    print("3. Analizar canal")
    print("4. Eliminar ultimo video")
    print("5. Salir")
    print("============================")

    opcion = int(input("Selecciona una opcion: "))

    # OPCION 1
    if opcion == 1:

        nombre = input("Nombre del video: ")
        videos.append(nombre)

        print("Video agregado correctamente")

    # OPCION 2
    elif opcion == 2:

        if videos == []:
            print("No hay videos en el canal")

        else:

            print("\n+--------------------------------------+")
            print("| ID | VIDEO                           |")
            print("+--------------------------------------+")

            i = 0

            while i < len(videos):

                print("|", i + 1, " | ", videos[i])

                i = i + 1

            print("+--------------------------------------+")

    # OPCION 3
    elif opcion == 3:

        cantidad = len(videos)

        print("\nANALISIS DE IA DEL CANAL")

        if cantidad == 0:
            print("Tu canal esta vacio")
            print("La IA recomienda subir videos")

        elif cantidad >= 1 and cantidad <= 3:
            print("Canal principiante")
            print("La IA recomienda mas constancia")

        elif cantidad >= 4 and cantidad <= 6:
            print("Canal en crecimiento")
            print("La IA detecta potencial")

        else:
            print("Canal exitoso")
            print("La IA detecta un futuro influencer")

    # OPCION 4
    elif opcion == 4:

        if videos == []:
            print("No hay videos para eliminar")

        else:
            eliminado = videos.pop()
            print("Video eliminado:", eliminado)

    # OPCION 5
    elif opcion == 5:

        print("Cerrando IA de Youtuber...")
        print("Hasta luego")

    # OPCION INVALIDA
    else:

        print("Opcion incorrecta")
```
